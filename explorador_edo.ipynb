# Explorador: EDO con SymPy

## Modelo

Usaremos la ley de enfriamiento de Newton:

$$\frac{dT}{dt}=-k(T-T_a), \qquad T(0)=T_0.$$

La meta es obtener una fórmula para $T(t)$.

```python
import sympy as sp

t, k, Ta, T0 = sp.symbols('t k Ta T0', positive=True)
T = sp.Function('T')

edo = sp.Eq(sp.diff(T(t), t), -k * (T(t) - Ta))
edo
```

## Resolver la EDO simbólicamente

`dsolve` busca una solución exacta. La condición inicial indica el valor de la función cuando $t=0$.

```python
solucion = sp.dsolve(edo, ics={T(0): T0})
solucion
```

## Reemplazar parámetros

`subs` cambia los símbolos por valores. `evalf` muestra una aproximación decimal.

```python
parametros = {k: 0.2, Ta: 20, T0: 90}
formula = solucion.rhs.subs(parametros)
formula
```

```python
for tiempo in [0, 5, 10]:
    temperatura = formula.subs(t, tiempo).evalf()
    print(f't = {tiempo:2}: T = {temperatura}')
```

## Verificar la condición inicial

Una solución debe cumplir el valor inicial que usamos para definir el problema.

```python
valor_inicial = solucion.rhs.subs(t, 0)
sp.simplify(valor_inicial - T0)
```

Si la última celda devuelve `0`, la fórmula cumple $T(0)=T_0$.
