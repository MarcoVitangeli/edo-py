# Playground: EDO con Python

Este cuaderno compara una solución simbólica con SymPy y una aproximación numérica con SciPy. Ejecuta cada celda con `Shift + Enter`.

## Modelo

Usaremos la ley de enfriamiento de Newton:

$$\frac{dT}{dt}=-k(T-T_a), \qquad T(0)=T_0.$$

```python
import sympy as sp

t, k, Ta, T0 = sp.symbols('t k Ta T0', positive=True)
T = sp.Function('T')

edo = sp.Eq(sp.diff(T(t), t), -k * (T(t) - Ta))
solucion_simbolica = sp.dsolve(edo, ics={T(0): T0})
solucion_simbolica
```

## Evaluar la solución simbólica

Cambia los valores de `k`, `Ta` y `T0` para explorar distintos escenarios.

```python
import numpy as np

parametros = {k: 0.2, Ta: 20, T0: 90}
T_simbolica = solucion_simbolica.rhs.subs(parametros)
modelo = sp.lambdify(t, T_simbolica, 'numpy')

for tiempo in [0, 5, 10]:
    print(f't = {tiempo:2}: T = {modelo(tiempo):.2f}')
```

## Resolver numéricamente con SciPy

`solve_ivp` aproxima la trayectoria aun cuando no tenemos una fórmula cerrada.

```python
from scipy.integrate import solve_ivp

def enfriamiento(t, y, k=0.2, Ta=20):
    return [-k * (y[0] - Ta)]

tiempos = np.linspace(0, 20, 200)
resultado = solve_ivp(enfriamiento, (0, 20), [90], t_eval=tiempos)
resultado.success
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
plt.plot(resultado.t, resultado.y[0], label='solución numérica')
plt.plot(tiempos, modelo(tiempos), '--', label='solución simbólica')
plt.axhline(20, color='gray', linestyle=':', label='temperatura ambiente')
plt.xlabel('tiempo')
plt.ylabel('temperatura')
plt.legend()
plt.grid(alpha=0.25)
plt.show()
```
