# Variación de parámetros con SymPy

Resolver $y''-2y'+y=e^x/(1+x^2)$, con $y(0)=1$ y $y'(0)=0$. Ejecuta las celdas en orden con `Shift + Enter`.

```python
import sympy as sp

x = sp.symbols('x', real=True)
y = sp.Function('y')
g = sp.exp(x)/(1 + x**2)
edo = sp.Eq(sp.diff(y(x), x, 2) - 2*sp.diff(y(x), x) + y(x), g)
edo
```

## Calcular los parámetros con el wronskiano

La homogénea tiene raíz doble $r=1$: $y_1=e^x$, $y_2=xe^x$. Usamos directamente $u_1=-\int y_2g/W\,dx$ y $u_2=\int y_1g/W\,dx$, con $W=y_1y_2'-y_1'y_2$. Las constantes de integración se incorporan a la solución homogénea.

```python
y1, y2 = sp.exp(x), x*sp.exp(x)
W = sp.simplify(sp.wronskian([y1, y2], x))
u1 = sp.integrate(sp.simplify(-y2*g/W), x)
u2 = sp.integrate(sp.simplify(y1*g/W), x)
{'W': W, 'u1': u1, 'u2': u2}
```

```python
yp = sp.simplify(u1*y1 + u2*y2)
yp
```

## Resolver con condiciones iniciales

`hint` especifica variación de parámetros y `ics` indica los dos datos iniciales.

```python
solucion = sp.dsolve(
    edo,
    hint='nth_linear_constant_coeff_variation_of_parameters',
    ics={y(0): 1, sp.diff(y(x), x).subs(x, 0): 0},
)
solucion
```

## Verificar el resultado

Sustituimos en la ecuación y en ambos datos iniciales. Los tres residuos deben ser cero.

```python
formula = solucion.rhs
residuos = {
    'EDO': sp.simplify(sp.diff(formula, x, 2) - 2*sp.diff(formula, x) + formula - g),
    'y(0) - 1': sp.simplify(formula.subs(x, 0) - 1),
    "y'(0)": sp.simplify(sp.diff(formula, x).subs(x, 0)),
}
assert all(valor == 0 for valor in residuos.values())
residuos
```
